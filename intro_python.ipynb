### Comparison & Logical Operators

These are the building blocks inside `if` conditions.

| Operator | Meaning              | Example          |
| -------- | -------------------- | ---------------- |
| `==`     | equal to             | `score == 90`    |
| `!=`     | not equal to         | `name != "Bob"` |
| `>`      | greater than         | `age > 18`       |
| `<`      | less than            | `gpa < 2.0`      |
| `and`    | both must be true    | `x > 0 and x < 10` |
| `or`     | at least one is true | `x == 0 or x == 1` |
| `not`    | flips True/False     | `not is_student` |

## 12. Functions — Great for reusing/streamlining the code.

````python
def greet(name):
    return f"Hello, {name}!"
````

````python
greet("Data Scientist") # we are passing this within the greet function
````

### `greet` is the function here. `name` is the parameter that we update when calling `greet`. It gets substituted in and the result is returned.

### Default Parameters

You can give a parameter a **default value** so it works even if the caller doesn't provide it.

````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

greet("Alice")               # uses the default → "Hello, Alice!"
# greet("Bob", "Hi there")  # custom greeting → "Hi there, Bob!"
````

### A practical function example

````python
def letter_grade(score):
    """Return a letter grade given a numeric score."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

print(letter_grade(95))  # A
print(letter_grade(72))  # C
````

## 15. Matplotlib — Visualizing Data
Matplotlib is Python's most widely used plotting library. The pyplot module (aliased as plt) is what you'll use 90% of the time.

`plt.` always needs the `x` and `y` parameters to work. Take it this as in coordinates as to where to place the data point.

````python
import matplotlib.pyplot as plt

gpas  = [3.8, 3.25, 3.6, 2.9, 3.1]
names = ["Jane", "Jack", "Alice", "Bob", "Charlie"]

plt.bar(names, gpas, color="steelblue") # bar plot
plt.title("Student GPAs")
plt.xlabel("Student")
plt.ylabel("GPA")
plt.ylim(0, 4.0)
plt.show()
````

````python
# Line plot — great for showing trends over time
years  = [2020, 2021, 2022, 2023, 2024]
values = [10, 15, 13, 20, 25]

plt.plot(years, values, marker="o", color="steelblue")
plt.title("Trend Over Time")
plt.xlabel("Year")
plt.ylabel("Value")
plt.show()
````

The main plot types:
- `plt.bar()` for bar charts,
- `plt.plot()` for line charts,
- `plt.scatter()` for scatter plots,
- `plt.hist()` for histograms

**plt.show()** always goes at the end to render the figure. Choose the appropriate plot to represent your data accurately!!

### 15b. Scatter Plots & Controlling Marker Size

`plt.scatter()` draws individual points instead of a line. A key feature is the `s=` parameter, which controls the **size** of each marker — this lets you encode a third variable (like population) through symbol size.

| Parameter | What it does |
|-----------|-------------|
| `x=`, `y=` | Coordinates of each point |
| `s=` | Marker area in points² — larger value = bigger dot |
| `alpha=` | Transparency: 0 = invisible, 1 = solid |
| `color=` | Marker fill color |

````python
# Simple scatter plot with variable marker sizes
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]
sizes = [50, 500, 100, 1000, 200]  # different size for each point

plt.scatter(x, y, s=sizes, color="steelblue", alpha=0.7)
plt.title("Scatter Plot with Variable Marker Size")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
````

#### Scaling sizes proportionally

When real-world values (like populations in the millions) are used directly as `s=`, the dots become enormous. A common trick is to **scale** the values down to a sensible dot-size range using division or a list comprehension.

````python
unscaled = [50000, 10000, 200000] # values are too big, need to normalize!

# Divide to bring into a sensible range for plotting
scaled = [i / 10000 for i in unscaled]  # list comprehension
````

#### A simple population choropleth map

````python
import matplotlib.pyplot as plt
````

`column` here is an input parameter of what column list to plot. Check table to see nomenclature of the column names!

Other parameters include:

- `legend` takes boolean like True/False
- `cmap` is for color scheme. Check different color maps https://matplotlib.org/stable/users/explain/colors/colormaps.html
- `figsize` which determines the overall figure size. Example: (15,10); 15 length, 10 height.

Each of these parameters can be seperated by `,` within the `.plot` function.

````python
ushapes.plot(column='POP_SQMI') # POP_SQMI is just population density which is population/area
plt.title('') # fill in with text
plt.xlabel('')
plt.ylabel('')
plt.show()
````

###### Notice how the higher values are not being represented accurately. This happens because of **outliers**. We need to remove them in order to have a better visual perspective.

#### Usually a good idea to remove extreme outliers in larger datasets to improve visual clarity such that higher values are represented correctly.

````python
# Calculate the 97th percentile for population density
cap_value = ushapes['POP_SQMI'].quantile(0.97)

# Cap the POP_SQMI values at the 97th percentile
ushapes['POP_SQMI_CAPPED'] = ushapes['POP_SQMI'].clip(upper=cap_value) # CLIPPING out the outliers
````

We have essentially added a new column called `POP_SQMI_CAPPED` in ushapes geodataframe with capped population density values from `POP_SQMI`.

###### Notice the difference now!

````python
# Outlier Removed Population Density Map
ushapes.plot(column="POP_SQMI_CAPPED", legend=True)
````

## 17. Lists, Dictionaries conversion to DataFrame

So far we have seen a **dictionary of lists** (columns-first). Another very common pattern is a **list of dictionaries**, where each dictionary represents one **record** (row). This is natural when each item has several attributes.

Think of it this way:
- **Dictionary of lists** → good for column-based storage (like a spreadsheet)
- **List of dictionaries** → good for record-based storage (like a table of rows)

````python
# Each fruit is one dictionary; all fruits collected in a list
fruits = [
    {"name": "Apple",   "color": "Red",     "taste": "Sweet"},
    {"name": "Banana",  "color": "Yellow",  "taste": "Sweet"},
    {"name": "Lemon",   "color": "Yellow",  "taste": "Sour"},
    {"name": "Blueberry", "color": "Blue", "taste": "Sweet-Sour"}
]

# Access the first record (index 0)
print(fruits[0])

# Access a specific field of the first record
print(fruits[0]["name"])
print(fruits[0]["color"])
````

#### Looping over a list of dictionaries

This is one of the most common patterns in data work — loop through all records and pull out individual fields.

````python
for i in fruits:
    print(f"{i['name']} tastes {i['taste']}")
````

#### Convert a list of dictionaries directly into a Pandas DataFrame

````python
import pandas as pd

fruits_df = pd.DataFrame(fruits)
fruits_df
````

### You can proceed to plot the data after converting to DataFrame.

**Note:** This is just an example plot, `scatter` not the accurate way to represent this dataset.

````python
plt.scatter(fruits_df["name"], fruits_df["color"]) # you are reading the columns as a list, see Part 15b.

# Add text labels to each point, we loop over the dataframe and use names, color as coordinates to print the name
for i in range(len(fruits_df)):
    plt.text(fruits_df["name"][i], fruits_df["color"][i], fruits_df["name"][i])

plt.show()
````
